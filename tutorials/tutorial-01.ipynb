```python
import io, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')

# Upload your own CSV in Google Colab
try:
    from google.colab import files
    uploaded = files.upload()
    csvs = [n for n in uploaded if n.lower().endswith('.csv')]
    if csvs:
        df = pd.read_csv(io.BytesIO(uploaded[csvs[0]]))
        print('Loaded:', csvs[0])
    else:
        print('No CSV selected; using a tiny demonstration dataset.')
        df = pd.DataFrame({'src_ip':['10.0.0.1','10.0.0.2','10.0.0.1','10.0.0.3','10.0.0.2'], 'failed_logins':[1,8,2,0,12], 'bytes':[500,900,650,300,1200], 'label':[0,1,0,0,1]})
except ImportError:
    df = pd.DataFrame({'src_ip':['10.0.0.1','10.0.0.2','10.0.0.1','10.0.0.3','10.0.0.2'], 'failed_logins':[1,8,2,0,12], 'bytes':[500,900,650,300,1200], 'label':[0,1,0,0,1]})

print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
display(df.head())
print('\nMissing values:\n', df.isna().sum())
display(df.describe(include='all').T)

numeric = df.select_dtypes(include=np.number)
if not numeric.empty:
    numeric.hist(figsize=(10,6), bins=20)
    plt.tight_layout(); plt.show()

if 'failed_logins' in df.columns:
    cutoff = df['failed_logins'].quantile(.90)
    print('Records at or above 90th percentile:')
    display(df[df['failed_logins'] >= cutoff].sort_values('failed_logins', ascending=False))

print('STUDENT TASKS')
print('1. Identify relevant columns for your cybersecurity question.')
print('2. Check data quality and class balance if a label exists.')
print('3. Adapt the example to a column in your selected dataset.')
print('4. Write two findings and one limitation.')
```
